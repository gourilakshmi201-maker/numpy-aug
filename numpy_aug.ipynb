```python
import numpy as np

#create a 4x4 matrix

z=np.array([[1,2,3,4],[3,2,1,5],[9,8,7,6],[4,5,3,2]])

print(z)
print(z.ndim)
print(z.size)
print(z.shape)

print(z[0])

print(z[2])

print(z[1,1])

print(z[0:,1])  #2nd coloum

print(z[0:,1:3])

print(z[2,1:3])

print(z[1:3,1:3])
```

```python
#zero matrix

z=np.zeros([3,3],dtype=int)

print(z)
```

```python
#one matrix

z=np.ones([3,3], dtype=complex)

print(z)
```

```python
#create two 2x2 matrix and store it in two variables

a=np.array([[1,2],[3,4]])

print(a)

b=np.array([[5,6],[7,8]])

print(b)

print(a+b)

print(a-b)

print(a*b)  #not matrix multiplication

print(a@b)

print(a.dot(b))
```

```python
#full matrix

z=np.full([3,3],4)

print(z)
```

```python
#identity matrix

z=np.identity(3)
print(z)

z=np.eye(4,dtype=int)
print(z)
```

```python
#arange

z=np.arange(1,10)
print(z)
```

```python
#create  a 4x3 matrix

z=np.ones([4,3])

print(z)

```

```python
z=np.ones([4,3]).reshape(6,2)
print(z)
```

```python
#create a 3x3 matrix with arange and reshape

z=np.arange(1,10).reshape(3,3)
print(z)


```

```python
#create a 4x3 matrix with float values

z=np.array([[3.1,4.5,7.7,3.5],[2.9,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])

print(z)

print(np.floor(z))

print(np.ceil(z))

print(np.round(z))  #nearest even number
```

```python
z=np.array([[3.1,4.5,7.7,3.5],[2.9,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])

print(z.max())
print(z.min())
print(z.sum())
```

```python
z=np.array([[3.1,4.5,7.7,3.5],[2.9,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])

print(z.max(axis=1))

print(z.max(axis=0))

print(np.sqrt(z))

print(np.sin(z))
print(np.cos(z))
```

```python
z=np.array([[3.1,4.5,7.7,3.5],[2.9,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])
print(np.sort(z))

print(np.sort(z,axis=1))

print(np.argsort(z,axis=1))
```
